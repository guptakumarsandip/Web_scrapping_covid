```python
from bs4 import BeautifulSoup
import requests

url = BeautifulSoup('https://www.worldometers.info/coronavirus/', 'html.parser')
#print(url)
soup = requests.get(url)
soup
```

```python
soup = soup.text
```

```python
#print(soup)
```

```python
!pip install lxml
```

```python
code = BeautifulSoup(soup, "lxml")
```

```python
#print(code)
```

```python
code = code.table
```

```python
#print(code)
```

```python
tags = code.find_all('tr')
```

```python
#print(tags)
```

```python
data = []
for tag in tags:
    x=tag.text
    data.append(x)
```

```python
#print(data)
```

```python
for i in data:
    i.split('\n')
```

```python
datas = []
for i in data:
    datas.append(i.split('\n')[1:-1])

#print(datas)
```

```python
import csv 
file = open('covid_data.csv','w')
x = csv.writer(file)
for i in datas:
    x.writerow(i)

file.close()
```

```python
cleaned_data = [i for i in datas if i[0] != '']
#print(cleaned_data)
```

```python
import csv 
file = open('covid_data.csv','w')
x = csv.writer(file)
for i in cleaned_data:
    x.writerow(i)

file.close()
```

```python
import pandas as pd
```

```python
import chardet

# Detect the encoding of the file
with open('covid_data.csv', 'rb') as file:
    result = chardet.detect(file.read())

# Use the detected encoding when reading the CSV file
df = pd.read_csv('covid_data.csv', encoding=result['encoding'])
```
